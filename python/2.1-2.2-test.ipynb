
### Assignment 3: if-elif-else Statement

Write a program that asks the user to input a number and prints whether the number is positive, negative, or zero.

```python
n = int(input("Podaj liczbę, żeby sprawdzić czy jest dodatnia"))
if(n>0):
    print(f"Liczba {n} jest dodatnia")
elif(n==0):
    print(f"Liczba {n} to zero")
else:
    print(f"Liczba {n} nie jest dodatnia")
```

### Assignment 4: Nested if Statement

Write a program that asks the user to input a number and prints whether the number is positive and even, positive and odd, or negative.

```python
n = int(input("Podaj liczbę"))
if(n>0):
    if(n%2==0):
        print(f"Liczba {n} jest dodatnia i jest parzysta")
    else:
        print(f"Liczba {n} jest dodatnia i jest nieparzysta")
elif(n==0):
    print(f"Liczba {n} to zero")
else:
    print(f"Liczba {n} nie jest dodatnia")
```

### Assignment 12: Factorial Calculation

Write a program that calculates the factorial of a number input by the user using a while loop.

```python
n = int(input("Podaj liczbę, której trzeba policzyć silnię"))
silnia=1
i=1
while(i<=n):
    silnia=silnia*i
    i=i+1
print(silnia)
```

### Assignment 13: Sum of Digits

Write a program that calculates the sum of the digits of a number input by the user using a while loop.

```python
n = int(input("Podaj liczbe: "))
suma_liczb = 0
while n > 0:
    liczba = n % 10
    suma_liczb += liczba
    n = n // 10
print(f"Suma liczb wynosi {suma_liczb}.")
```

### Assignment 14: Prime Number Check

Write a program that checks if a number input by the user is a prime number using a for loop.

```python
n = int(input("Podaj liczbę: "))
czy_pierwsza = True
if n <= 1:
    czy_pierwsza = False
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            czy_pierwsza = False
            break
if czy_pierwsza:
    print(f"{n} jest liczbą pierwszą.")
else:
    print(f"{n} nie jest liczbą pierwszą.")
```
